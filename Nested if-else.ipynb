```python
#6.wap to find middle element is even or odd

s=[3,4,6,7,9,1,5]
mi = len(s) // 2
me = s[mi]

print(f"Middle element is: {me}")

# Check if it's even or odd
if me % 2 == 0:
    print("The middle element is even.")
else:
    print("The middle element is odd.")
```

```python
#7.wap to purchase a phone from the shopping app

apps = ['flipkart', 'amazon']
categories = ['electronics', 'mobile', 'fashion', 'furnitures']

print("Available apps:", apps)
print("Available categories:", categories)

select_app = input('Enter the app you want to use: ')

if select_app in apps:
    print(f'You selected {select_app}')

    select_c = input('Select a category: ')

    if select_c == 'mobile':
        mobiles = ['vivo', 'realme']
        print("Available mobiles:", mobiles)

        sm = input('Select the mobile you want to buy: ')

        if sm == 'vivo':
            print('Vivo mobile price is ₹10,000')
            c1 = input("Confirm the purchase (type 'yes' to buy, 'no' to cancel): ")
            if c1 == 'yes':
                print(' Purchase completed! You bought a Vivo.')
            else:
                print(' You declined the payment.')

        elif sm == 'realme':
            print('Realme mobile price is ₹12,000')
            c1 = input("Confirm the purchase (type 'yes' to buy, 'no' to cancel): ")
            if c1 == 'yes':
                print(' Purchase completed! You bought a Realme.')
            else:
                print(' You declined the payment.')
        else:
            print('Selected mobile not available.')
    else:
        print('Only "mobile" category is supported for purchase right now.')
else:
    print('Invalid app selection.')
```

```python
#8.wap to give 10% off only who is purchasing in credit card and min 3 product should purchase and each product price should be more than 500



pm = input("Enter payment method (credit card / debit card / cash): ")

np = int(input("Enter number of products you want to purchase: "))

pp = []
for i in range(np):
    price = float(input(f"Enter price of product {i+1}: ₹"))
    pp.append(price)

if pm == "credit card" and np >= 3 and all(price > 500 for price in pp):
    total = sum(pp)
    discount = total * 0.10
    fp = total - discount
    print(f"\n You got a 10% discount!")
    print(f"Original Total: ₹{total}")
    print(f"Discount: ₹{discount}")
    print(f"Final Amount to Pay: ₹{fp}")
else:
    print("\n No discount applicable.")
    print(f"Total Amount: ₹{sum(pp)}")
```
