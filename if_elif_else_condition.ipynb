```python
#1.wap to check whether the given character is uppercase/lowercase/digit/special (with and without using inbuilt function)

a = eval(input('Enter the character :'))

if a.isupper():
    print(f'{a} is Upper case')

elif a.islower():
    print(f'{a} is lower case')

elif a.isdigit():
    print(f'{a} is digit')

else:
    print('it is special character')
```

```python
#1.wap to check whether the given character is uppercase/lowercase/digit/special (with and without using inbuilt function)

a = eval(input('Enter the character :'))
b = ord(a)
if b >= 97 and b <= 122: 
    print(f'{a} is lower case')

elif b>=65 and b<=90:
    print(f'{a} is upper case')
elif b>=48 and b<=57:
    print(f'{a} is digit')

else:
    print('it is special character')
```
