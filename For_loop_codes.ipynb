```python
#6.wap to print the count of alphabets and numbers and space in the given string

s="india got the independence in the year 1947"
space = 0
alpha = 0
digit = 0
for i in s:
    if i.isspace():
        space = space + 1
    elif i.isdigit():
        digit = digit+1

    elif i.isalpha():
        alpha = alpha +1

print(f'total space are {space}')

print(f'total alphabets are {alpha}')
print(f'total digits are {digit}')

```

```python
#9.wap to create a dictionary and traverse into it and if the length is even print as it else reverse it

#output:-->{'apple': 'elppa', 'google': 'google', 'yahoo': 'oohay', 'microsoft': 'tfosorcim', 'gmail': 'liamg', 'walmart': 'tramlaw'}



names=["apple","google","yahoo","microsoft","gmail","walmart"]
d = {}
for i in names:
    if len(i) % 2 == 0:
        d[i] = i
        

    else:
        d[i] = i[::-1]
print(d)
```

```python
#22.wap to create a dictionary with letter and its words starting with that letter pair

#o/p:-->{'h': ['hi', 'hello'], 'g': ['good'], 'm': ['morning'], 'w': ['welcome'], 't': ['to'], 'p': ['python'], 's': ['session']}

s="hi hello good morning welcome to python session"

s = "hi hello good morning welcome to python session"
words = s.split()
letter_words_dict = {}

for word in words:
    first_letter = word[0].lower()  # Convert to lowercase for consistency
    if first_letter in letter_words_dict:
        letter_words_dict[first_letter].append(word)
    else:
        letter_words_dict[first_letter] = [word]

print(letter_words_dict)

#chatgpt
    
```

```python
#23.wap to create a dictionary of characters and its indices pair
s="hello python"


#o/p:-->{"h":[0,9],"e":1..........}

s = "hello python"
char_indices = {}

for index, char in enumerate(s):
    if char in char_indices:
        char_indices[char].append(index)
    else:
        char_indices[char] = [index]

print(char_indices)

#chatgpts

```
