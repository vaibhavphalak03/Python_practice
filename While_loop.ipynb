```python
#16.wap if a names ends with vowels then reverse a names else print its length
names=["Kumar","Lakita","Umesh a","Priyanka"]

i = 0

while i < len(names):
    if names[i][-1] in 'aeiou':
        print(names[i][::-1])

    else:
        print(len(names[i]))

    i+=1
        
```
